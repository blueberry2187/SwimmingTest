```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/KeithGalli/Olympics-Dataset/refs/heads/master/results/results.csv")
df

##Importing Dataset from the given URL using pandas read_csv function. The dataset contains Olympic results data, which can be analyzed and manipulated using pandas DataFrame functionalities.
```

```python
df.isnull().sum()
##Checking for missing values in the dataset using the isnull() method followed by sum(). This will help identify any columns that have null values, which may need to be handled before further analysis.
```

```python
df.duplicated().sum()
##Checking for duplicate rows in the dataset using the duplicated() method followed by sum(). This will help identify if there are any repeated entries that may need to be removed to ensure data integrity.
```

```python
df.drop_duplicates()
##Removing duplicate rows from the dataset using the drop_duplicates() method. This will help ensure that the dataset contains only unique entries, which is important for accurate analysis and results.
```

```python
df= df[df["Discipline"] == "Swimming (Aquatics)"]
df
```

```python
for i in df["Event"].unique():
    print(i)
#We use this over the two list to identify water polo and diving events, which we want to remove from the dataset. By printing the unique values in the "Event" column, we can see all the different swimming events and filter out any that are not relevant to our analysis.
```

```python
eventset = {"100 metres Freestyle, Men (Olympic)"}
for i in df["Event"].unique():
    inclEvent = True
    if (i == "200 metres Obstacle Course, Men (Olympic)" or i == "Water Polo, Men (Olympic)"):
        inclEvent = False
    if (i == "Underwater Swimming, Men (Olympic)" or i == "Plunge For Distance, Men (Olympic)"):
        inclEvent = False
    if (i == "Discus Throw, Men (Olympic)" or i == "Football, Men (Olympic)" or i == "Shot Put, Men (Olympic)" or i == "Standing High Jump, Men (Olympic)"):
        inclEvent = False
    if (not i.endswith("(Olympic)")):
        inclEvent = False
    if (inclEvent):
        eventset.add(i)
eventset
##Creating a list of all of our events that we want to look at within our dataset. We are excluding events that are not relevant to swimming, such as obstacle courses, water polo, and other non-swimming events. The final set of events will be used for further analysis.
```

```python
df = df[(df["Event"] != "200 metres Obstacle Course, Men (Olympic)") & (df["Event"] != "Water Polo, Men (Olympic)") & (df["Event"] != "Underwater Swimming, Men (Olympic)") & (df["Event"] != "Plunge For Distance, Men (Olympic)") & (df["Event"] != "Discus Throw, Men (Olympic)") & (df["Event"] != "Football, Men (Olympic)") & (df["Event"] != "Shot Put, Men (Olympic)") & (df["Event"] != "Standing High Jump, Men (Olympic)") & (df["Event"].str.endswith("(Olympic)"))]
df
#The actual code that filters out the non-swimming aquatic events from the dataset. It uses boolean indexing to exclude rows where the "Event" column matches any of the specified non-swimming events or does not end with "(Olympic)". This ensures that the resulting DataFrame contains only relevant swimming events for further analysis.
```

```python
filtered_Events = df["Event"].unique().tolist()
filtered_Events
#Confirming that all the events in our new dataset are relevant to swimming and that we have successfully filtered out any non-swimming events. This will help ensure that our analysis is focused on the appropriate data.
```
