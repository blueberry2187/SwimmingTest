Checkpoint 2 - Data Visualizations

```python
import pandas as pd
import matplotlib.pyplot as plt
import squarify
df = pd.read_csv("https://raw.githubusercontent.com/KeithGalli/Olympics-Dataset/refs/heads/master/results/results.csv")
df.drop_duplicates()
df= df[df["Discipline"] == "Swimming (Aquatics)"]
df = df[(df["Event"] != "200 metres Obstacle Course, Men (Olympic)") & (df["Event"] != "Water Polo, Men (Olympic)") & (df["Event"] != "Underwater Swimming, Men (Olympic)") & (df["Event"] != "Plunge For Distance, Men (Olympic)") & (df["Event"] != "Discus Throw, Men (Olympic)") & (df["Event"] != "Football, Men (Olympic)") & (df["Event"] != "Shot Put, Men (Olympic)") & (df["Event"] != "Standing High Jump, Men (Olympic)") & (df["Event"].str.endswith("(Olympic)"))]
df_gold = df[df["Medal"] == "Gold"]
df_silver = df[df["Medal"] == "Silver"]
df_bronze = df[df["Medal"] == "Bronze"]
df_male = df[df["Event"].str.contains("Men", na=False, regex=False)]
df_female = df[df["Event"].str.contains("Women", na=False, regex=False)]
df_male_gold = df_male[df_male["Medal"] == "Gold"]
df_male_silver = df_male[df_male["Medal"] == "Silver"]
df_male_bronze = df_male[df_male["Medal"] == "Bronze"]
df_female_gold = df_female[df_female["Medal"] == "Gold"]
df_female_silver = df_female[df_female["Medal"] == "Silver"]
df_female_bronze = df_female[df_female["Medal"] == "Bronze"]
df
```

```python
df_noc_agg = df.groupby("NOC").agg("count").reset_index()
df_noc_agg_male = df_male.groupby("NOC").agg("count").reset_index()
df_noc_agg_female = df_female.groupby("NOC").agg("count").reset_index()
df_noc_agg
#Aggregating the events by country to check the total medals by each country, male and female
```

```python
df_noc_agg_gold = df_gold.groupby("NOC").agg("count").reset_index()
df_noc_agg_male_gold = df_male_gold.groupby("NOC").agg("count").reset_index()
df_noc_agg_female_gold = df_female_gold.groupby("NOC").agg("count").reset_index()
df_noc_agg_gold
#Aggregating the events by country for gold medal-winning performances to check the total gold medals by each country.
```

```python
df_noc_agg_silver = df_silver.groupby("NOC").agg("count").reset_index()
df_noc_agg_male_silver = df_male_silver.groupby("NOC").agg("count").reset_index()
df_noc_agg_female_silver = df_female_silver.groupby("NOC").agg("count").reset_index()
df_noc_agg_silver
#Aggregating the events by country for silver medal-winning performances to check the total silver medals by each country.
```

```python
df_noc_agg_bronze = df_bronze.groupby("NOC").agg("count").reset_index()
df_noc_agg_male_bronze = df_male_bronze.groupby("NOC").agg("count").reset_index()
df_noc_agg_female_bronze = df_female_bronze.groupby("NOC").agg("count").reset_index()
df_noc_agg_bronze
#Aggregating the events by country for bronze medal-winning performances to check the total bronze medals by each country.
```

1. LOOKING AT THE MEDALS USA HAS GAINED AS BAR GRAPH AND LINE GRAPH - Felix Pilny

```python
df_noc_agg_gold_10 = df_noc_agg_gold.nlargest(10, "Medal")
df_noc_agg_silver_10 = df_noc_agg_silver.nlargest(10, "Medal")
df_noc_agg_bronze_10 = df_noc_agg_bronze.nlargest(10, "Medal")
df_noc_agg_10 = df_noc_agg.nlargest(10, "Medal")
df_noc_agg_10
#Looking at top 10 countries with the most gold, silver, and bronze medals in swimming events.
```

```python
df_noc_agg_gold = df_noc_agg_gold.sort_values(by="Medal", ascending=False)
df_noc_agg_gold.plot.bar(x="NOC", y="Medal", title="Gold Medalists in Swimming Events",figsize=(15, 3), color="gold")
#Graphing every country that has recieved gold medals in swimming events.
```

```python
plt.figure(figsize=(15,3))
squarify.plot(sizes=df_noc_agg_gold['Medal'], label=df_noc_agg_gold['NOC'], alpha=.8)
plt.axis('off')
plt.show()

##Making a treemmap of every country that has recieved gold medals in swimming events. The size of the box is determined by the number of gold medals won by each country.
```

```python
plt.figure(figsize=(15,3))
squarify.plot(sizes=df_noc_agg_gold_10['Medal'], label=df_noc_agg_gold_10['NOC'], alpha=.8)
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(15,3))
squarify.plot(sizes=df_noc_agg_silver_10['Medal'], label=df_noc_agg_silver_10['NOC'], alpha=.8)
plt.axis('off')
plt.show()
```
