Checkpoint 1 - Data Finalized and Cleaned

```python
import pandas as pd
df = pd.read_csv("https://raw.githubusercontent.com/KeithGalli/Olympics-Dataset/refs/heads/master/results/results.csv")
df.drop_duplicates()
df= df[df["Discipline"] == "Swimming (Aquatics)"]
df = df[(df["Event"] != "200 metres Obstacle Course, Men (Olympic)") & (df["Event"] != "Water Polo, Men (Olympic)") & (df["Event"] != "Underwater Swimming, Men (Olympic)") & (df["Event"] != "Plunge For Distance, Men (Olympic)") & (df["Event"] != "Discus Throw, Men (Olympic)") & (df["Event"] != "Football, Men (Olympic)") & (df["Event"] != "Shot Put, Men (Olympic)") & (df["Event"] != "Standing High Jump, Men (Olympic)") & (df["Event"].str.endswith("(Olympic)"))]
df
```

```python
df_NON_Agg = df.groupby("NOC").agg("count").reset_index()
df_NON_Agg
#Aggregating the events by country to check the total medals by each country
```

```python
df_gold = df[df["Medal"] == "Gold"]
df_gold
#Creating a DataFrame called df_gold that contains only the rows where the "Medal" column is equal to "Gold". This will allow us to focus our analysis on gold medal-winning performances in swimming events.
```

```python
df_silver = df[df["Medal"] == "Silver"]
df_silver
#Creating a DataFrame called df_silver that contains only the rows where the "Medal" column is equal to "Silver". This will allow us to focus our analysis on silver medal-winning performances in swimming events.
```

```python
df_bronze = df[df["Medal"] == "Bronze"]
df_bronze
#Creating a DataFrame called df_bronze that contains only the rows where the "Medal" column is equal to "Bronze". This will allow us to focus our analysis on bronze medal-winning performances in swimming events.
```

```python
df_NOC_Agg_Gold = df_gold.groupby("NOC").agg("count").reset_index()
df_NOC_Agg_Gold
#Aggregating the events by country for gold medal-winning performances to check the total gold medals by each country.
```

```python
df_NOC_Agg_Silver = df_silver.groupby("NOC").agg("count").reset_index()
df_NOC_Agg_Silver
#Aggregating the events by country for silver medal-winning performances to check the total silver medals by each country.
```

```python
df_NOC_Agg_Bronze = df_bronze.groupby("NOC").agg("count").reset_index()
df_NOC_Agg_Bronze
#Aggregating the events by country for bronze medal-winning performances to check the total bronze medals by each country.
```
